## Q6. Median value of horsepower

1. Find the median value of the `horsepower` column.
2. Calculate the most frequent value (mode) of `horsepower`.
3. Use `fillna` to fill the missing values in `horsepower` with that mode.
4. Calculate the median again.

Has it changed? (increased / decreased / no)

*Hint:* `mode()` returns a Series (there can be ties); take the first element with `[0]`. Think: does `median()` ignore NaN?

```python
median_before = df['horsepower'].median()
print(median_before)
mode = df['horsepower'].mode()[0]
print(mode)
fill_horsepower = df['horsepower'].fillna(mode)
median_after = fill_horsepower.median()
print(median_after)      
```

## Q7. Sum of weights

1. Select all the cars from Asia.
2. Select only the columns `vehicle_weight` and `model_year`.
3. Select the first 7 values.
4. Get the underlying NumPy array: `X`.
5. Compute `XTX = X.T @ X`.
6. Invert `XTX`.
7. Create `y = [1100, 1300, 800, 900, 1000, 1100, 1200]`.
8. Compute `w = inv(XTX) @ X.T @ y`.
9. What's the sum of all the elements of `w`?

*Hints:* `.values` (or `.to_numpy()`) gives the array; `np.linalg.inv` inverts; `@` is matrix multiplication. Check `X.shape`, `XTX.shape` and `w.shape` at each step: if the shapes don't make sense, something is wrong.

```python
df_asia_selection = df_asia[['vehicle_weight', 'model_year']]
```

```python
print(df_asia_selection.head(7))
```

```python
numpy_asia_values = df_asia_selection.head(7).values
```

```python
numpy_asia_values.shape
```

```python
X = numpy_asia_values

XTX = X.T @ X
print(XTX.shape)

XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])   

w = XTX_inv @ X.T @ y     
print(w)                  
print(w.sum())            
```
